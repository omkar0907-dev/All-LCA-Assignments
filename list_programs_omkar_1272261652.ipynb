```python
# Program 4 : Write a program to count occurence of an element in the list

nums = [10, 20, 10, 30, 10, 40, 50]
element = int(input("Enter number to search :"))
count = nums.count(element)
print(f"{element} occurs {count} number of times in the list")
```

```python
# Program 4 (optimised) : Write a program to count occurence of an element in the list

nums = [10, 20, 10, 30, 10, 40, 50]
freq = {}
for i in range(len(nums)):
    if freq.get(nums[i], -1) == -1:
        freq[nums[i]] = 1
    else:
        freq[nums[i]] += 1

element = int(input("Enter number to search :"))
if freq.get(element, -1) == -1:
    print(f"{element} NOT found in list")
else:
    print(f"{element} occurs {freq[element]} number of times in list")
```
