```python
# Progam 1 : Write a program to add an item to the dictionary
_dict = {0 : 10, 1 : 20}
print(f"Dictionary before adding item : {_dict}")

_dict[2] = 30
print(f"Dictionary after adding item : {_dict}")
```

```python
# Program 2 : Write a program to concatenate the following dictionaries to create a new one

d1 = {1 : 10, 2 : 20}
d2 = {3 : 30, 4 : 40}
d3 = {5: 50, 6 : 60}

_dict = {}

_dict.update(d1)
_dict.update(d2)
_dict.update(d3)

print(f"Final _dict : {_dict}")
```

```python
# Progam 3 : Write a program to check if a given key already exist in a dictionary

_dict = {101 : "Amit", 102 : "Rahul", 103 : "Priya"}
key = int(input("Enter key to search : "))
if key in _dict:
    print(f"{key} exists in the dictionary")
else:
    print(f"{key} does NOT exist in the dictionary")
```

```python
# Progam 4 : Write a program to iterate ovar a dictionary using for loop and print keys alone, values alone and keys and values together
_dict = {101 : "Amit", 102 : "Rahul", 103 : "Priya"}
for key in _dict:
    print(key)
for key in _dict:
    print(_dict[key])
for key, value in _dict.items():
    print(key, value)
```

```python
# Program 5 : Write a program to sum all the values in a dictionary, considering the values will be of int type
_dict = {"a" : 10, "b" : 20, "c" : 30, "d" : 40}
total = 0
for val in _dict.values():
    total += val
print(f"Sum of values : {total}")
```

```python
# Progam 5 : Write a program to0 replace the last of tuples in a list to 100
nums = [(10, 20, 40), (40, 50, 60), (70, 80, 90)]
new_list = []

for tup in nums:
    new_tup = tup[:-1] + (100,)
    new_list.append(new_tup)

print(f"Original List : {nums}")
print(f"Updated list : {new_list}")
```

```python
# Progam 6 : Write a program to find the maximum and minimum elment of a tuple
nums = (25, 10, 45, 5, 60, 30)
_max = max(nums)
_min = min(nums)
print(f"Maximum of {nums} is {_max}")
print(f"Minimum of {nums} is {_min}")
```

```python
# Python program to swap the first and last element of a tuple
nums = (10, 20, 30, 40, 50)
new_tuple = (nums[-1],) + nums[1:-1] + (nums[0],)
print(f"Original tuple : {nums}")
print(f"Tuple after swaping : {new_tuple}")
```
