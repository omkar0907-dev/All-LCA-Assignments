```python
# Matrix addition using lists

A = [
    [1, 2, 3],
    [4, 5, 6]
]

B = [
    [7, 8, 9],
    [10, 11, 12]
]

# Create an empty matrix
C = [
    [0, 0, 0],
    [0, 0, 0]
]

# Add corresponding elements
for i in range(len(A)):
    for j in range(len(A[0])):
        C[i][j] = A[i][j] + B[i][j]

print("Matrix A:")
for row in A:
    print(row)

print("Matrix B:")
for row in B:
    print(row)

print("Sum of matrices:")
for row in C:
    print(row)
```

```python
import numpy as np

# Create two matrices
A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

B = np.array([
    [7, 8, 9],
    [10, 11, 12]
])

# Add the matrices
C = A + B

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)

print("\nSum of matrices:")
print(C)
```
